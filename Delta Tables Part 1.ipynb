```sql
--Temporary Views --> gets removed once the session is ended
CREATE TEMPORARY VIEW sales_temp_vw
AS
SELECT * FROM catalog_s.bronze.sales
WHERE InvoiceNo ='536365'
```

```sql
desc extended sales_temp_vw 
```

```sql
--permanent view 
CREATE VIEW catalog_s.bronze.sales_vw
AS
SELECT * FROM catalog_s.bronze.sales
WHERE InvoiceNo ='536365'
```

```sql
CREATE OR REPLACE VIEW catalog_s.bronze.sales_vw
AS
SELECT * FROM catalog_s.bronze.sales
where InvoiceNo = '536367'
```

```sql
select * from catalog_s.bronze.sales_vw
```

```sql
--CTAS Delta Table
CREATE TABLE catalog_s.bronze.sales_ctas
AS
SELECT * from catalog_s.bronze.sales
```

```sql
select * from catalog_s.bronze.sales_ctas
```

```sql
describe extended catalog_s.bronze.sales_ctas
```

```sql
--DEEP CLONE -- copies both data and metadata
create table catalog_s.bronze.sales_dc
DEEP CLONE catalog_s.bronze.sales
```

```sql
--SHALLOW CLONE -- copies only metadata and data is still pointing to the old location

create table catalog_s.bronze.sales_sc
SHALLOW CLONE catalog_s.bronze.sales
```
