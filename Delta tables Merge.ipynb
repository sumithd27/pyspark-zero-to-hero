```sql
--merge data to update existing records and insert new records
MERGE into catalog_s.bronze.emp e
USING catalog_s.bronze.emp_updates u
ON e.emp_id = u.emp_id
WHEN MATCHED THEN
UPDATE SET 
e.salary=u.salary,
e.is_active='Y'
WHEN NOT MATCHED THEN
INSERT (
  emp_id,
  emp_name,
  dept_code,
  salary,
  is_active
) VALUES (
  u.emp_id,
  u.emp_name,
  u.dept_code,
  u.salary,
  'Y'
)
WHEN NOT MATCHED BY SOURCE THEN 
UPDATE SET is_active ='N'
;
```
