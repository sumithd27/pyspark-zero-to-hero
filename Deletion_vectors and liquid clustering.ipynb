```sql
create schema if not exists catalog_s.bronze;
```

```sql
create table catalog_s.bronze.sales as 
select * from 
read_files('dbfs:/databricks-datasets/online_retail/data-001/data.csv',
          header=>true,
          format=>'csv'
          --inferSchema=True
          )
```

```sql
use catalog_s.bronze;
select Count(*) from sales
```

```sql
describe extended catalog_s.bronze.sales
```

```sql
--change the table property
alter table catalog_s.bronze.sales set tblproperties (delta.enableDeletionVectors= true);
```

```sql
-- Delete a record without deletion vectors enabled 
delete from catalog_s.bronze.sales where InvoiceNo = '540644'
```

```sql
describe history catalog_s.bronze.sales
```

```sql
-- Delete a record with deletion vectors enabled 
delete from catalog_s.bronze.sales where InvoiceNo = '536368'
```
